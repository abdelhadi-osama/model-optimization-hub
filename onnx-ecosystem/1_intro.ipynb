## 🧠 The Core Problems in Deep Learning Deployment - part 1

### 1️⃣ Framework Interoperability and Vendor Lock-in 🔗

Deep learning teams often collaborate using diverse training frameworks like **PyTorch**, **TensorFlow**, **Keras**, or **scikit-learn**. Historically, there has been no direct or easy way to translate a model trained in one framework into another for further integration. This creates isolated, closed pipelines where developers are locked into specific ecosystems, hindering team collaboration and slowing down development. 🚧



### 2️⃣ Hardware Fragmentation 🖥️⚙️

In the deployment phase, models must run on a massive variety of edge hardware, including **Raspberry Pi**, **FPGAs**, **Intel CPUs**, and **NVIDIA GPUs**. Without a universal standard, deploying a model requires using the specific software pipeline dictated by the hardware manufacturer (such as **OpenVINO** for Intel or **TensorRT** for NVIDIA). This forces engineering teams to manage a complex, fragmented web of custom conversions just to get their model running on different target devices. 🕸️



### 3️⃣ Resource Constraints on Edge Devices 🔋📉

Edge computing devices inherently possess limited computational power, memory capacity, and battery life compared to centralized cloud servers. Raw models exported directly from training frameworks are not built for this environment; they often contain redundant training-specific operators (like **Dropout**) and rely on heavy, high-precision floating-point numbers (**FP32**). Deploying these raw models directly results in slow execution 🐌, high latency ⏳, and excessive power consumption ⚡.

# 🧩 What Is ONNX? - part 2

**ONNX (Open Neural Network Exchange)** is an open-source, vendor-neutral standard format and intermediate representation for deep learning and machine learning models. Initially developed by **Facebook** and released collaboratively with **Microsoft** in **2017** 📅 Later on, other companies like AMD, Arm, Baidu, Huawei, HuggingFace, IBM, Intel, Mediatek, Nvidia, and Qualcomm have extended their support t for a common framework to accelerate development in the field of deep learning , ONNX provides a common abstraction that unifies model definitions across the AI ecosystem 🌐.

Here is an intuitive breakdown of the solutions ONNX provides: 👇



## 1️⃣ The Software Solution: Freedom from Framework Lock-in 🔓

- 🌍 **Universal Compatibility:** ONNX solves the problem of closed pipelines. A team can train a model in **PyTorch**, **TensorFlow**, **Keras**, or **Scikit-Learn**, and easily convert it into a single, standardized ONNX file.

- 🤝 **Seamless Team Collaboration:** Because the format is universal, engineers do not have to waste time manually translating code from one framework to another. An engineer who specializes in **TensorFlow** can easily pick up and integrate a model that was originally built in **PyTorch**.

- 💻 **Multi-Language Support:** Real-world applications are rarely written purely in Python. ONNX solves this by providing native support for languages like **C++**, **C#**, and **Java**, allowing the model to be easily plugged into existing software systems.



## 2️⃣ The Hardware Solution: "Train Once, Deploy Anywhere" 🚀

- 🔌 **Hardware Independence:** Instead of being forced to use a specific hardware manufacturer's custom deployment tool, you can use the ONNX ecosystem to deploy your model universally.

- ⚙️ **The ONNX Runtime Engine:** Through **ONNX Runtime**, the exact same model file can be deployed across a massive variety of hardware targets, including **Intel CPUs**, **NVIDIA GPUs**, **AMD GPUs**, **Android phones**, **FPGAs**, and **Raspberry Pi**. This drastically simplifies the engineering effort required to scale a product across different devices.



## 3️⃣ The Optimization Solution: Making Models Edge-Ready 🪶

- 🧹 **Cleaning Up the Math:** Models coming fresh out of training are often bloated with steps that are only useful for the training process. Tools in the ONNX ecosystem (like **ONNX Simplifier** and **ONNX Optimizer**) act like a cleanup crew, automatically finding and removing these redundant steps to speed up execution.

- 📦 **Compressing for Efficiency:** ONNX Runtime can shrink the overall size of the model (a process called **quantization**). This makes the model light enough to run quickly and efficiently on edge devices that have very limited memory and battery power.

# ☁️ Cloud computing VS Edge computing 🔌

## 1. The Core Difference 
The fundamental distinction between cloud and edge computing is based entirely on where the data is actually processed. While the data used for processing is usually generated by sensors, it can also originate from user inputs or software applications, such as a user interacting with a locally executed chatbot on a mobile phone.

* **☁️ Cloud Computing:** Data is sent over a network to a centralized server equipped with higher computation capabilities. This server processes large chunks of the data, and the results are then either stored in the cloud or returned to the source.


* **🔌 Edge Computing:** Computation is performed at or near the source of the data to reduce the overall latency of the execution. Instead of sending information to a distant server, the processing is handled locally by devices such as smartphones, IoT devices, or dedicated edge hardware.



## 2. Why We Need Edge Computing (The Benefits)

You correctly identified the main reasons we move computation to the edge. By keeping processing local, we gain several advantages:

* **⚡ Reduced Latency (Speed):** Because the data does not need to travel to a distant server and back, transfer time is minimized, enabling instant, real-time execution.


* **📶 Bandwidth Efficiency:** Processing data locally drastically reduces the volume of information sent over the network, lowering overall bandwidth requirements and network load.


* **💰 Cost Efficiency:** Minimizing the amount of data sent to the cloud inherently reduces the operational costs associated with data transfer and cloud storage.


* **🔒 Better Privacy and Security:** Because sensitive data is generated, processed, and stored locally, it is not transmitted across the internet, which safeguards it from interception and misuse.



## 3. The Limitations of Edge Computing

While edge computing solves network and speed issues, it introduces hardware and logistical challenges:

* **🧠 Limited Computation Power:** Edge devices inherently possess less processing power and storage capacity than centralized cloud infrastructure, which limits the complexity of the AI models or tasks they can handle.


* **🛡️ Physical Safety and Security:** Because edge devices are deployed in the real world rather than in secure data centers, they are highly vulnerable to physical tampering, unauthorized access, and targeted sabotage.


* **🔧 Scalability and Maintenance:** Managing devices spread across massive geographical locations is logistically difficult; rolling out regular software updates, applying fixes, and scaling the network requires manual effort and increases operational costs.

# 🧠 Deep Learning Model Optimization for Edge Deployment


Deep learning models are usually trained on GPUs hosted on cloud or in-premise. Once models have been trained, they undergo the following steps before they are ready for deployment on edge devices:

### ✂️ Model Pruning :
 is an algorithmic optimization technique that identifies and removes redundant, less important weights from a trained neural network to make it more lightweight and efficient for deployment. While the core goal is to reduce the model's parameter count with minimal impact on accuracy, its effect on the network's physical structure depends on the specific approach taken.

 ### 🎓  Distillation
When a large model is pruned to make it smaller and faster, it often loses some of its original accuracy because it is working with fewer parameters. Instead of retraining this smaller model entirely from scratch, we pair it with the original, highly accurate large model.

* **👨‍🏫 The Teacher:** The large model acts as a "Teacher."
* **🎓 The Student:** The small, pruned model acts as a "Student."

During training, the student doesn't just try to learn from the raw dataset; it actively tries to mimic the exact predictions and behaviors of the teacher model. This process successfully transfers the deep "knowledge" of the heavy model into the lightweight model, resulting in a small model that is much more accurate than it would be if trained on its own.

### ⚙️ Graph Optimizations

Once the trained model is obtained it might contain many redundant nodes and parameters that can be either removed from the graph or fused together with other nodes and parameters.

The trained graph would undergo several optimizations such as **Constant Folding**, **NoOp Removal**, **Dead Code Elimination**, and **Kernel Fusion**, which can make it more efficient during execution.

### 🔢 Quantization

Quantization is an optimization technique used to compress a deep learning model so it runs faster and uses less power. It works by shrinking the precision of the numbers the model uses to perform calculations, typically converting complex decimals (like 32-bit floating-point numbers) into simpler, smaller whole numbers (like 8-bit integers).

Because computer processors can handle basic integer math much faster and with less effort than complicated decimals, this conversion significantly speeds up the model's execution time and reduces its physical file size by up to four times.

While trading high-precision math for these simpler integer numbers can introduce minor computational errors that slightly lower the model's overall accuracy, the massive gains in speed, memory efficiency, and energy savings make quantization an essential step for deploying models on small, resource-constrained edge devices.

# Overview of the ONNX Model Architecture
onnx uses Protocol Buffers (Protobuf) as its serialization format. Protobuf is language-independent, meaning that different programming languages can read and work with the same serialized model.

The main structure of an ONNX model is called ModelProto.

Think of ModelProto as a container that holds the model's information, computational graph, and reusable functions.

![alt text](img/model_arch_onnx.png "My image")

## 📦 ModelProto — The Main Container

`ModelProto` contains the main information needed to describe the ONNX model.

#### 🔑 Important Fields

| Component | What does it tell us? |
|---|---|
| **IR Version** | Which version of the ONNX model representation are we using? |
| **Opset Imports** | Which versions of the operator definitions are we using? |
| **GraphProto** | How is the model's computation organized? |
| **FunctionProto** | Which reusable computational functions are defined? |

#### 📝 Remember

- 🏷️ **IR Version** → Version of the ONNX **representation**
- ⚙️ **Opset Version** → Version of the **operator specifications**
- 🕸️ **GraphProto** → The **computational graph**
- 🔁 **FunctionProto** → **Reusable computational logic**

## 🕸️ GraphProto — The Computational Graph

This is where we describe how the model performs its computations.

Imagine a neural network containing:

Input → Convolution → ReLU → MaxPool → Output


`GraphProto` describes the operations, their connections, and the tensors flowing between them.

It contains several important structures:


## 🔧 A. NodeProto — What Operations Do We Perform?

A **node** represents one operation in the computational graph.

### 📋 Examples

For example:
- 🔹 `Conv`
- 🔹 `ReLU`
- 🔹 `MatMul`
- 🔹 `Add`
- 🔹 `MaxPool`

### 🔍 Each `NodeProto` Describes

| Field | Purpose |
|---|---|
| **Operation Type** | Which operation should happen? |
| **Inputs** | Which tensor values enter the operation? |
| **Outputs** | Which tensor values does it produce? |
| **Attributes** | What settings does the operation use? |

### ⚙️ AttributeProto

`AttributeProto` describes the **configuration** of an operation.

For example, a convolution node might have:
- 📏 Strides
- 🧱 Padding
- 🔍 Dilations
- 🗂️ Groups

> 💡 Think of `NodeProto` as the **instructions** for processing data.


### Note 
- **Dilations** 🕳️ → Spacing between kernel elements (inserts gaps in the filter so it "sees" a wider area without more weights). One-liner: *how far apart the kernel's sampling points are.*

- **Groups** 🗂️ → Splits input channels into groups, each convolved separately (e.g. `groups=1` = normal conv, `groups=channels` = depthwise conv). One-liner: *how many independent channel groups the convolution is split into.*

```python
import onnx
model = onnx.load("resnet18-v1-7.onnx")

# Onnx models are stored as ModelProto object.
print("Onnx model stored as: ", type(model))
```

```python
# Each ONNX model would have a particular Opset version and IR version.
print("Model IR version: ", model.ir_version)
print("Model opset version: ", model.opset_import[0].version)
```

```python
# Each ModelProto object would contain several GraphProto objects.
graph = model.graph
print("Graph stored as: ", type(graph))

# Each GraphProto object contains NodeProto objects.
print(f"Total nodes in graph: {len(graph.node)}\n")

first_conv_node = graph.node[0]
print(f"First convolution node: \n\n{first_conv_node}")
```

## 📊 B. TensorProto — Where Is the Tensor Data?

A **tensor** represents multidimensional data used by the model.

`TensorProto` describes and can store the **actual tensor data**.

### 🔍 It Includes

| Information | Example |
|---|---|
| **Tensor Name** | `conv_weights` |
| **Dimensions** | `[32, 3, 3, 3]` |
| **Data Type** | `FLOAT`, `INT64` |
| **Tensor Values** | Actual numerical values |

### 💾 Tensor Values Can Represent

- 🧠 Learned weights
- ⚖️ Biases
- 📌 Constants

> 💡 Think of `TensorProto` as the **data** used by the computational operations.

```python
# Model's weights are stored as constant nodes or initializers.
# Initializers are stored as TensorProto inside GraphProto.
'''
The TensorProto are stored in the "initializer"
 property of the ModelProto. The TensorProto are referenced as input in the NodeProto.
'''
print("Total constants in graph: ", len(graph.initializer))
print("Initializer type: ", type(graph.initializer[0]))
print("\nFirst initializer: \n\n", graph.initializer[0])
```

```python
# Actual tensor value from initializer can be obtained as below.

first_init = graph.initializer[0]
first_init_value = onnx.numpy_helper.to_array(first_init)
print("First initializer")
print("name: ", first_init.name)
print("shape: ", first_init_value.shape) 
print("dtype: ", first_init_value.dtype)
print("value: ", first_init_value)
```

## 🏷️ C. ValueInfoProto — What Are the Tensor Characteristics?

`ValueInfoProto` describes **information about tensor values** flowing through the graph.

### 🔍 It Includes

- 📛 Tensor **Name**
- 🔢 **Data Type**
- 📐 **Shape**

### 📋 For Example
Name: input
Data Type: FLOAT
Shape: [1, 3, 640, 640]


It tells us **what kind of tensor** we are working with — **not** the actual numerical values inside it.

### 🧠 Remember the Difference

| Proto | What it holds |
|---|---|
| 📊 `TensorProto` | Tensor **data and values** |
| 🏷️ `ValueInfoProto` | Tensor **description and characteristics** |


### Note :
The shape and datatype information is important from an optimization standpoint. If any operator is doing redundant work, then it can be identified via the shape and datatype information. For example, a Cast node converts a given tensor’s data type into boolean. If the given input itself is Boolean, then the cast node is redundant.

```python
# Each node's input and output tensors have shape and data type 
# information. This info is stored as ValueInfoProto structure of GraphProto.

all_value_infos = model.graph.value_info
print("Total number of value info : ", len(all_value_infos))

# There are no value info available in graph because intermediate nodes 
# doesn't have any shape information. This information can be obtained by 
# performing shape inference.
model = onnx.shape_inference.infer_shapes(model)

all_value_infos = model.graph.value_info
print("Total number of value info : ", len(all_value_infos))

first_tensor_val_info = all_value_infos[0]
print("Value Info type: ", type(first_tensor_val_info))
print("\nFirst tensor's value info: ", first_tensor_val_info)
```

## 🔁 FunctionProto — Reusable Computational Functions

`FunctionProto` describes **reusable computational logic** made from ONNX nodes.

### 🧩 Imagine This

Your model repeatedly uses this sequence:
MatMul → Add → ReLU


Instead of defining the same sequence repeatedly, you can define it **once as a function** and reference it from the graph.

### 📦 A `FunctionProto` Contains

- 📛 Function **name** and **domain**
- 🔌 **Input** and **output** names
- 🔧 `NodeProto` definitions
- ⚙️ **Attributes**
- 📚 **Opset imports**
- 🏷️ Optional `ValueInfoProto` information

> 💡 Think of `FunctionProto` as a **reusable collection of computational instructions**.

### 🛠️ Use Case — Custom Operators

It can be used to define **custom operators** within a model.

For example, consider a case where a user needs to develop functionality for a new activation function **PReLU**, assuming that ONNX does not support it. The user can then define the **PReLU operator via `FunctionProto`** to support such models.
